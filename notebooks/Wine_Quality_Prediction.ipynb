## 4.6 Detecting Outliers Using the Interquartile Range (IQR) Method

## Objective

The objective of this stage is to identify potential outliers using the Interquartile Range (IQR) statistical method.

The IQR method identifies observations that fall significantly below or above the typical range of the data.

Using both statistical analysis and visual inspection provides greater confidence in identifying unusual observations.

````python
#
...
# SELECT NUMERICAL COLUMNS
#
...

numerical_columns = wine_clean.select_dtypes(include={"number"})

numerical_columns.head()
````

````python
#
...
# CALCULATE IQR
#
...

Q1 = numerical_columns.quantile(0.25)
Q3 = numerical_columns.quantile(0.75)

IQR = Q3 - Q1

print("First Quartile (Q1)")
print(Q1)

print("Third Quartile (Q3)")
print(Q3)

print("Interquartile Range (IQR)")
print(IQR)
````

````python
#
...
# CALCULATE OUTLIER LIMITS
#
...

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Lower Bound")
print(lower_bound)

print("\nUpper Bound")
print(upper_bound)
````

````python
#
...
# COUNT OUTLIERS
#
...

outlier_count = (
    (numerical_columns < lower_bound) |
    (numerical_columns > upper_bound)
).sum()

print("Number of potential outliers in each column:")
print(outlier_count)
````

## Interpretation

The Interquartile Range (IQR) method was used to identify potential outliers across all numerical variables.

The results confirmed that some chemical properties contain values outside the normal statistical range.

The boxplots highlighted visually prominent extreme values in some variables, particularly **free sulfur dioxide** and **total sulfur dioxide**.

However, the IQR analysis showed that these variables did not have the highest number of statistical outliers. Instead, **fixed acidity, volatile acidity, chlorides, quality,** and **sulphates** contained the greatest number of potential outliers.

Owing to the fact that these measurements may represent genuine characeristics of certain wines rather than data entry errors, the potential outliers were retained for subsequent analysis.

This demonstrates the importance of combining visual inspection with statistical methods when identifying outliers.

## Business Insight

Using the IQR method provides an objective statistical approach to identifying unusual observations.

Retaining valid extreme values allows the machine learning model to learn from diverse range of wine characteristics while maintaining the integrity of the dataset.

## 6.4 Splitting the Dataset into Training and Testing Sets

Before training a machine learning model, it is important to divide the dataset into two separate parts: the training set and the testing set.

The training set is used to teach the model the relationships between the input variables and the target variable, while the testing set is used to evaluate how well the trained model performs on new, unseen data.

This process helps to make sure that the model generalizes well and prevents overly optimistic performance estimates.

### Why Did I Split the Dataset?

If the same data is used for both training and testing, the model may simply memorize the answers instead of learning meaningful patterns. This results in poor performance when making predictions on new data.

To avoid this problem, the tataset is split into:

- **Training Set (80%)** - Used to train the model.
- **Testing Set (20%)** - Used to evaluate the model.

This is a standard practice in supervised machine learning.

````python
#
...
# IMPORT TRAIN-TEST SPLIT
#
...

from sklearn.model_selection import train_test_split
````

````python
#
...
# SPLIT DATA INTO TRAINING AND TESTING SETS
#
...

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size = 0.20,
    random_state = 42
)

print("Training Features:", X_train.shape)
print("Testing Features:", X_test.shape)
print("Training Targets:", y_train.shape)
print("Test Targets:", y_test.shape)

````

## Interpretation of Results

The dataset has been successfully divided into two subsets:

* The **training set** contains **80%** of the data and will be used to train the machine learning model.

* The **testing set** contains the remaining **20%** and will be used to evaluate the model's ability to make predictions on unseen data.

The **random_state=42** parameter ensures that the split is reproducible, meaning anyone running the notebook will obtain the same training and testing sets.

## Business Insight

Separating the dataset into training and testing sets helps ensure that the machine leaning model performs well not only on historical data but also on new wine samples.
